```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import drive

# --- Configuración ---
drive.mount("/content/gdrive/")

DATA_DIR = Path(
    "/content/gdrive/MyDrive/01_UNSa_Carrera/4to_Año/"
    "Metodos_Cuantitativos/TrabajoFinal/datasets"
)

# Escenario -> nombre del archivo (ajustar a tus nombres reales)
ARCHIVOS = {
    "S0": "datasetS0.csv",
    "S1": "datasetS1.csv",
    "S1_alt": "datasetS1alt.csv",
    "S2": "datasetS2.csv",
    "S3": "datasetS3.csv",
    "S4": "datasetS4.csv",
    "S5": "datasetS5.csv",
    "S6": "datasetS6.csv",
}


# --- Funciones ---
def cargar_escenario(nombre: str) -> pd.DataFrame:
    """Lee el CSV de un escenario y agrega la columna 'escenario'."""
    ruta = DATA_DIR / ARCHIVOS[nombre]
    df = pd.read_csv(ruta)
    df["escenario"] = nombre
    return df


def cargar_todos() -> pd.DataFrame:
    """Concatena todos los escenarios disponibles en un solo DataFrame."""
    return pd.concat(
        [cargar_escenario(n) for n in ARCHIVOS], ignore_index=True
    )


from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import drive

# --- Configuración ---
drive.mount("/content/gdrive/")

DATA_DIR = Path(
    "/content/gdrive/MyDrive/01_UNSa_Carrera/4to_Año/"
    "Metodos_Cuantitativos/TrabajoFinal/datasets"
)

# Escenario -> nombre del archivo (ajustar a tus nombres reales)
ARCHIVOS = {
    "S0": "datasetS0.csv",
    "S1": "datasetS1.csv",
    "S1_alt": "datasetS1alt.csv",
    "S2": "datasetS2.csv",
    "S3": "datasetS3.csv",
    "S4": "datasetS4.csv",
    "S5": "datasetS5.csv",
    "S6": "datasetS6.csv",
}


# --- Funciones ---
def cargar_escenario(nombre: str) -> pd.DataFrame:
    """Lee el CSV de un escenario y agrega la columna 'escenario'."""
    ruta = DATA_DIR / ARCHIVOS[nombre]
    df = pd.read_csv(ruta)
    df["escenario"] = nombre
    return df


def cargar_todos() -> pd.DataFrame:
    """Concatena todos los escenarios disponibles en un solo DataFrame."""
    return pd.concat(
        [cargar_escenario(n) for n in ARCHIVOS], ignore_index=True
    )


columnas = {n: set(pd.read_csv(DATA_DIR / f, nrows=0).columns)
            for n, f in ARCHIVOS.items()}
comunes = set.intersection(*columnas.values())
print("Columnas comunes:", len(comunes))
for n, c in columnas.items():
    print(n, len(c), "| faltan respecto a la unión:", len(set.union(*columnas.values()) - c))
```

```python
extra = set(pd.read_csv(DATA_DIR / ARCHIVOS["S6"], nrows=0).columns) - comunes
print(extra)

s6 = cargar_escenario("S6")
col = list(extra)[0]
print(s6[col].describe())
print(s6[col].isna().mean())            # proporción de vacíos
print(s6.groupby("attack_active")[col].describe())  # ¿cambia con el ataque?
```

```python
df = pd.concat([cargar_escenario(n) for n in ARCHIVOS], ignore_index=True)

def grupo(c):
    return c.split(".", 1)[0] if "." in c else "tiempo/etiqueta"

num = {c: pd.api.types.is_numeric_dtype(df[c]) for c in df.columns}

dic = pd.DataFrame({
    "grupo": [grupo(c) for c in df.columns],
    "tipo": df.dtypes.astype(str).values,
    "pct_vacios": (df.isna().mean() * 100).round(2).values,
    "min": [df[c].min() if num[c] else None for c in df.columns],
    "max": [df[c].max() if num[c] else None for c in df.columns],
    "ejemplo": [df[c].dropna().iloc[0] if df[c].notna().any() else None
                for c in df.columns],
}, index=df.columns)

dic.to_csv(DATA_DIR / "diccionario_borrador.csv")
print(dic.head(20))
```

```python
medidores = ["PV-Inverter-AC-Meter", "BSS-Inverter-AC-Meter", "Load-A-AC-Meter",
             "Load-B-AC-Meter", "Substation-AC-Meter"]
fases = ["pln1", "pln2", "pln3"]

# 1) ¿Qué es Substation realtime?
col = "Substation-AC-Meter.realtime"
print(df.groupby("escenario")[col].agg(["min", "max", "nunique"]))

# 2) Diferencia entre p_sum3 y la suma de fases, por escenario y ataque
resid = pd.DataFrame({
    m: df[f"{m}.p_sum3"] - df[[f"{m}.{f}" for f in fases]].sum(axis=1, min_count=3)
    for m in medidores
})
resid["escenario"] = df["escenario"]
resid["attack_active"] = df["attack_active"]
print(resid.groupby(["escenario", "attack_active"])
      .agg(lambda s: s.abs().mean()).round(1))

# 3) ¿En qué escenario están las 31 filas vacías de la batería?
cols_bss = [c for c in df.columns if c.startswith("BSS-Inverter-AC-Meter.")]
vacias = df[cols_bss].isna().all(axis=1)
print(df.loc[vacias, "escenario"].value_counts())
```

```python
#Filas de contexto
perf = [c for c in df.columns if c.startswith("Profile.")]
print(perf)
print(df[perf].describe().T)
print(df[perf].isna().mean().round(3))

# Filtrar solo columnas numéricas para calcular la media sin errores
perf_num = df[perf].select_dtypes(include='number').columns.tolist()

# ¿Cambian entre escenarios? (si sí, pueden delatar el escenario)
print(df.groupby("escenario")[perf_num].mean().round(2))

# ¿Cambian con el ataque dentro de un mismo escenario?
print(df.groupby(["escenario", "attack_active"])[perf_num].mean().round(2))
```

```python
perf = [c for c in df.columns if c.startswith("Profile.")]

# 196 filas sin perfil: ¿en qué escenario?
print(df[df["Profile.total_irradiance"].isna()]["escenario"].value_counts())

# ¿La irradiancia cae con el tiempo dentro de cada escenario?
df["t"] = pd.to_datetime(df["timestamp"], utc=True)
for esc, g in df.groupby("escenario"):
    g = g.dropna(subset=["Profile.total_irradiance"]).sort_values("t")
    seg = (g["t"] - g["t"].iloc[0]).dt.total_seconds()
    print(esc, round(seg.corr(g["Profile.total_irradiance"]), 2))
```
