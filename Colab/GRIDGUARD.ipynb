```python
import os
import pandas as pd

DATA_PATH = "/kaggle/input/datasets/stephenkamran/grid-guard/data4cyber_dataset"

scenario_folders = [
    "S0_benign_baseline",
    "S1_industroyer_pv",
    "S1_industroyer_pv_alt",
    "S2_industroyer_bss",
    "S3_arp_spoof_bss_meter_half_values",
    "S4_arp_spoof_loads_pv_two_phase",
    "S5_arp_spoof_loads_pv_bss_two_phase",
    "S6_mqtt_supply_chain_compromise"
]

frames = []

for scenario in scenario_folders:
    fp = os.path.join(DATA_PATH, scenario, "dataset.csv")
    temp = pd.read_csv(fp)

    # Preserve LOSO scenario identity
    temp["scenario"] = scenario

    # Numeric analysis label derived directly from authoritative attack_active
    temp["attack_binary"] = temp["attack_active"].astype(int)

    frames.append(temp)

master = pd.concat(frames, ignore_index=True, sort=False)

print("Master shape:", master.shape)

print("\nAttack distribution:")
print(master["attack_binary"].value_counts().sort_index())

print("\nMissing scenario labels:", master["scenario"].isna().sum())
print("Duplicate full rows:", master.duplicated().sum())

print("\nExpected:")
print("Rows    = 14354")
print("Columns = 161")
print("Benign  = 5880")
print("Attack  = 8474")
```

```python
import numpy as np
import pandas as pd

print("=== DATA TYPE COUNTS ===")
print(master.dtypes.value_counts())

print("\n=== NON-NUMERIC / BOOLEAN COLUMNS ===")
non_numeric = master.select_dtypes(exclude=[np.number]).columns.tolist()

for col in non_numeric:
    print(f"{col:45s} | {master[col].dtype}")

print("\nTotal non-numeric/boolean columns:", len(non_numeric))


# Explicitly known target / leakage / provenance fields
known_meta = [
    "attack_active",
    "attack_phase",
    "attack_phase_all",
    "scenario",
    "attack_binary",
    "Attacker.event",
    "Attacker.new_value",
    "Attacker.old_value",
    "MQTT-Price-Signal.event"
]

print("\n=== KNOWN TARGET / LEAKAGE / PROVENANCE FIELDS PRESENT ===")
for col in known_meta:
    if col in master.columns:
        print(f"✓ {col:35s} | dtype={master[col].dtype}")
    else:
        print(f"✗ {col}")


print("\n=== OTHER POSSIBLE METADATA COLUMNS ===")
for col in master.columns:
    lc = col.lower()

    # Do NOT flag operational phase_seq or realtime meter variables here
    if (
        "timestamp" in lc
        or lc in ["time", "datetime", "date"]
        or "scenario_id" in lc
        or "label" in lc
        or "target" in lc
    ):
        print(col)
```

```python
# Exact target / leakage / provenance exclusions
EXCLUDE_COLS = [
    "timestamp",
    "Profile.timestamp",
    "attack_active",
    "attack_phase",
    "attack_phase_all",
    "scenario",
    "attack_binary",
    "Attacker.event",
    "Attacker.new_value",
    "Attacker.old_value",
    "MQTT-Price-Signal.event"
]

# Check that every intended exclusion exists
missing_exclusions = [c for c in EXCLUDE_COLS if c not in master.columns]

print("Missing exclusions:", missing_exclusions)

# Candidate operational predictors
candidate_features = [
    c for c in master.columns
    if c not in EXCLUDE_COLS
]

print("\nTotal master columns:", master.shape[1])
print("Excluded columns:", len(EXCLUDE_COLS))
print("Candidate predictors:", len(candidate_features))

# Verify all candidates are numeric
non_numeric_candidates = [
    c for c in candidate_features
    if not pd.api.types.is_numeric_dtype(master[c])
]

print("\nNon-numeric candidate predictors:")
print(non_numeric_candidates)

print("\nExpected candidate predictor count = 150")
```

```python
# Step 12 — Identify zero-variance predictors

zero_variance_cols = []

for col in candidate_features:
    if master[col].nunique(dropna=False) <= 1:
        zero_variance_cols.append(col)

print("Zero-variance columns:")
for col in zero_variance_cols:
    print(" -", col)

print("\nZero-variance count:", len(zero_variance_cols))
print("Expected count = 9")
```

```python
# Step 13 — Create final strict predictor set

final_features = [
    c for c in candidate_features
    if c not in zero_variance_cols
]

X_final = master[final_features].copy()
y_final = master["attack_binary"].copy()

print("Final predictor count:", len(final_features))
print("X_final shape:", X_final.shape)
print("y_final shape:", y_final.shape)

print("\nRemaining Attacker.* columns:")
print([c for c in final_features if c.startswith("Attacker.")])

print("\nAny known leakage fields remaining:")
print([c for c in EXCLUDE_COLS if c in final_features])

print("\nNaN values in X_final:", X_final.isna().sum().sum())
print("Infinite values in X_final:",
      np.isinf(X_final.to_numpy(dtype=float)).sum())

print("\nExpected:")
print("Final predictors = 141")
print("X_final shape = (14354, 141)")
print("Remaining Attacker.* = []")
print("Known leakage remaining = []")
```

```python
# Step 14 — Save frozen strict analysis dataset

import os
import pandas as pd

OUTPUT_DIR = "/kaggle/working"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Final frozen analysis dataframe
strict_master = X_final.copy()

# Add only non-feature identifiers/target for later LOSO evaluation
strict_master["attack_binary"] = y_final.values
strict_master["scenario"] = master["scenario"].values

save_path = os.path.join(
    OUTPUT_DIR,
    "gridguard_strict_master_141_features.csv"
)

strict_master.to_csv(save_path, index=False)

print("Saved:", save_path)
print("Saved shape:", strict_master.shape)

print("\nColumns:")
print("141 predictors + attack_binary + scenario =", strict_master.shape[1])

print("\nAttack counts:")
print(strict_master["attack_binary"].value_counts().sort_index())

print("\nScenario counts:")
print(strict_master["scenario"].value_counts())
```

```python
# Step 15 — Create scenario-safe 20-step temporal windows

import numpy as np

WINDOW_SIZE = 20
STRIDE = 1

X_seq_parts = []
y_seq_parts = []
scenario_seq_parts = []

for scenario in scenario_folders:
    df_s = strict_master[strict_master["scenario"] == scenario].reset_index(drop=True)

    X_s = df_s[final_features].to_numpy(dtype=np.float32)
    y_s = df_s["attack_binary"].to_numpy(dtype=np.int8)

    X_windows = []
    y_windows = []

    for start in range(0, len(df_s) - WINDOW_SIZE + 1, STRIDE):
        end = start + WINDOW_SIZE

        X_windows.append(X_s[start:end])

        # Current-state label = label of final timestep
        y_windows.append(y_s[end - 1])

    X_windows = np.asarray(X_windows, dtype=np.float32)
    y_windows = np.asarray(y_windows, dtype=np.int8)

    X_seq_parts.append(X_windows)
    y_seq_parts.append(y_windows)

    scenario_seq_parts.append(
        np.repeat(scenario, len(y_windows))
    )

    print(
        f"{scenario:40s} "
        f"rows={len(df_s):4d}  "
        f"windows={len(y_windows):4d}"
    )

# Combine all scenario-safe windows
X_seq_strict = np.concatenate(X_seq_parts, axis=0)
y_seq_strict = np.concatenate(y_seq_parts, axis=0)
scenario_seq = np.concatenate(scenario_seq_parts, axis=0)

print("\n=== FINAL TEMPORAL DATASET ===")
print("X_seq_strict shape:", X_seq_strict.shape)
print("y_seq_strict shape:", y_seq_strict.shape)
print("scenario_seq shape:", scenario_seq.shape)

unique, counts = np.unique(y_seq_strict, return_counts=True)

print("\nWindow label distribution:")
for label, count in zip(unique, counts):
    print(f"{label}: {count}")

print("\nNaNs:", np.isnan(X_seq_strict).sum())
print("Infinite values:", np.isinf(X_seq_strict).sum())

print("\nExpected:")
print("X_seq_strict = (14202, 20, 141)")
print("Attack windows = 8455")
print("Benign windows = 5747")
```

```python
# Step 16 — Identify pure-benign windows for AE training

pure_benign_parts = []
pure_benign_scenarios = []

for scenario in scenario_folders:
    df_s = strict_master[strict_master["scenario"] == scenario].reset_index(drop=True)

    X_s = df_s[final_features].to_numpy(dtype=np.float32)
    y_s = df_s["attack_binary"].to_numpy(dtype=np.int8)

    scenario_pure = []

    for start in range(0, len(df_s) - WINDOW_SIZE + 1, STRIDE):
        end = start + WINDOW_SIZE

        # Pure benign means ALL 20 labels are benign
        if np.all(y_s[start:end] == 0):
            scenario_pure.append(X_s[start:end])

    scenario_pure = np.asarray(scenario_pure, dtype=np.float32)

    pure_benign_parts.append(scenario_pure)

    pure_benign_scenarios.extend(
        [scenario] * len(scenario_pure)
    )

    print(
        f"{scenario:40s} "
        f"pure-benign windows={len(scenario_pure):4d}"
    )

X_pure_benign = np.concatenate(pure_benign_parts, axis=0)
pure_benign_scenarios = np.asarray(pure_benign_scenarios)

print("\n=== PURE-BENIGN AE DATASET ===")
print("X_pure_benign shape:", X_pure_benign.shape)
print("Total pure-benign windows:", len(X_pure_benign))

print("\nNaNs:", np.isnan(X_pure_benign).sum())
print("Infinite values:", np.isinf(X_pure_benign).sum())

print("\nExpected:")
print("Pure-benign windows = 5561")
print("Shape = (5561, 20, 141)")
```

```python
# Step 16 — Identify pure-benign windows for AE training
# Correctly handles scenarios with zero benign windows

pure_benign_parts = []
pure_benign_scenarios = []

for scenario in scenario_folders:
    df_s = strict_master[
        strict_master["scenario"] == scenario
    ].reset_index(drop=True)

    X_s = df_s[final_features].to_numpy(dtype=np.float32)
    y_s = df_s["attack_binary"].to_numpy(dtype=np.int8)

    scenario_pure = []

    for start in range(0, len(df_s) - WINDOW_SIZE + 1, STRIDE):
        end = start + WINDOW_SIZE

        # Pure benign = ALL 20 timesteps benign
        if np.all(y_s[start:end] == 0):
            scenario_pure.append(X_s[start:end])

    # Handle scenarios with zero pure-benign windows
    if len(scenario_pure) > 0:
        scenario_pure = np.stack(
            scenario_pure,
            axis=0
        ).astype(np.float32)

        pure_benign_parts.append(scenario_pure)

        pure_benign_scenarios.extend(
            [scenario] * len(scenario_pure)
        )

        count = len(scenario_pure)

    else:
        count = 0

    print(
        f"{scenario:40s} "
        f"pure-benign windows={count:4d}"
    )

# Combine only non-empty scenario arrays
X_pure_benign = np.concatenate(
    pure_benign_parts,
    axis=0
)

pure_benign_scenarios = np.asarray(
    pure_benign_scenarios
)

print("\n=== PURE-BENIGN AE DATASET ===")
print("X_pure_benign shape:", X_pure_benign.shape)
print("Total pure-benign windows:", len(X_pure_benign))

print("\nNaNs:", np.isnan(X_pure_benign).sum())
print("Infinite values:", np.isinf(X_pure_benign).sum())

print("\nExpected:")
print("Pure-benign windows = 5561")
print("Shape = (5561, 20, 141)")
```

```python
# Step 18 — Strict Random Forest LOSO baseline

import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, recall_score, f1_score

RF_PARAMS = {
    "n_estimators": 120,
    "max_depth": 18,
    "min_samples_leaf": 2,
    "class_weight": "balanced",
    "random_state": 42,
    "n_jobs": -1
}

rf_results = []

for held_out in scenario_folders:

    # Row-level LOSO split
    train_mask = strict_master["scenario"] != held_out
    test_mask  = strict_master["scenario"] == held_out

    X_train = strict_master.loc[train_mask, final_features]
    y_train = strict_master.loc[train_mask, "attack_binary"]

    X_test = strict_master.loc[test_mask, final_features]
    y_test = strict_master.loc[test_mask, "attack_binary"]

    # Train RF
    rf = RandomForestClassifier(**RF_PARAMS)
    rf.fit(X_train, y_train)

    # Predict
    y_pred = rf.predict(X_test)

    # Basic metrics
    acc = accuracy_score(y_test, y_pred)

    # Attack recall / FSR
    if (y_test == 1).sum() > 0:
        attack_recall = recall_score(
            y_test,
            y_pred,
            pos_label=1,
            zero_division=0
        )

        fsr = 1.0 - attack_recall

        f1 = f1_score(
            y_test,
            y_pred,
            pos_label=1,
            zero_division=0
        )
    else:
        attack_recall = np.nan
        fsr = np.nan
        f1 = np.nan

    # False alarm rate on benign observations
    benign_mask = (y_test == 0)

    if benign_mask.sum() > 0:
        far = (
            (y_pred[benign_mask] == 1).sum()
            / benign_mask.sum()
        )
    else:
        far = np.nan

    rf_results.append({
        "scenario": held_out,
        "n_test": len(y_test),
        "accuracy": acc,
        "attack_recall": attack_recall,
        "FSR": fsr,
        "F1": f1,
        "FAR": far
    })

    print(f"\n=== {held_out} ===")
    print("Train shape:", X_train.shape)
    print("Test shape :", X_test.shape)
    print("Accuracy   :", round(acc, 4))
    print(
        "Attack Recall:",
        "N/A" if np.isnan(attack_recall)
        else round(attack_recall, 4)
    )
    print(
        "FSR:",
        "N/A" if np.isnan(fsr)
        else round(fsr, 4)
    )
    print(
        "F1:",
        "N/A" if np.isnan(f1)
        else round(f1, 4)
    )
    print(
        "FAR:",
        "N/A" if np.isnan(far)
        else round(far, 4)
    )


rf_results_df = pd.DataFrame(rf_results)

print("\n\n=== RF LOSO RESULTS ===")
print(rf_results_df.to_string(index=False))

# Mean across attack scenarios only
attack_only = rf_results_df[
    rf_results_df["attack_recall"].notna()
]

print("\n=== ATTACK-SCENARIO MEAN ===")
print("Mean accuracy :", round(attack_only["accuracy"].mean(), 4))
print("Mean recall   :", round(attack_only["attack_recall"].mean(), 4))
print("Mean FSR      :", round(attack_only["FSR"].mean(), 4))
print("Mean F1       :", round(attack_only["F1"].mean(), 4))
```

```python
# Step 22 — Define the exact GRIDGUARD-AI TCN architecture
# No training yet.

import tensorflow as tf
from tensorflow.keras import layers, Model

WINDOW_SIZE = 20
N_FEATURES = 141


def residual_tcn_block(x, filters, dilation_rate, dropout_rate=0.20):
    """
    Residual causal TCN block:
    Causal Conv1D -> BN -> ReLU -> Dropout -> Residual Add -> ReLU
    """

    residual = x

    x = layers.Conv1D(
        filters=filters,
        kernel_size=3,
        padding="causal",
        dilation_rate=dilation_rate
    )(x)

    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    x = layers.Dropout(dropout_rate)(x)

    # Residual projection when channel dimensions differ
    if residual.shape[-1] != filters:
        residual = layers.Conv1D(
            filters=filters,
            kernel_size=1,
            padding="same"
        )(residual)

    x = layers.Add()([x, residual])
    x = layers.ReLU()(x)

    return x


def build_gridguard_tcn(input_shape=(20, 141)):

    inputs = layers.Input(
        shape=input_shape,
        name="grid_telemetry"
    )

    # Block 1: 32 filters, dilation 1
    x = residual_tcn_block(
        inputs,
        filters=32,
        dilation_rate=1
    )

    # Block 2: 32 filters, dilation 2
    x = residual_tcn_block(
        x,
        filters=32,
        dilation_rate=2
    )

    # Block 3: 64 filters, dilation 4
    x = residual_tcn_block(
        x,
        filters=64,
        dilation_rate=4
    )

    # Temporal aggregation
    x = layers.GlobalAveragePooling1D()(x)

    # Dense decision layer
    x = layers.Dense(
        32,
        activation="relu"
    )(x)

    x = layers.Dropout(0.30)(x)

    outputs = layers.Dense(
        1,
        activation="sigmoid",
        name="attack_probability"
    )(x)

    model = Model(
        inputs=inputs,
        outputs=outputs,
        name="GRIDGUARD_TCN"
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=1e-3
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


# Build only for verification
tcn_check = build_gridguard_tcn(
    input_shape=(WINDOW_SIZE, N_FEATURES)
)

tcn_check.summary()

print("\nInput shape :", tcn_check.input_shape)
print("Output shape:", tcn_check.output_shape)
```

```python
# Step 25 — Rebuild and lock canonical TCN configuration
# NO TRAINING YET

import os
import random
import numpy as np
import tensorflow as tf

from tensorflow.keras import layers, Model

# --------------------------------------------------
# Canonical training configuration
# --------------------------------------------------

SEEDS = [11, 42, 77]

BATCH_SIZE = 64
MAX_EPOCHS = 30

LEARNING_RATE = 1e-3

EARLY_STOP_PATIENCE = 5

LR_REDUCE_FACTOR = 0.5
LR_REDUCE_PATIENCE = 2
MIN_LR = 1e-5


# --------------------------------------------------
# Deterministic seed helper
# --------------------------------------------------

def set_global_seed(seed):

    os.environ["PYTHONHASHSEED"] = str(seed)

    random.seed(seed)
    np.random.seed(seed)

    tf.keras.utils.set_random_seed(seed)

    try:
        tf.config.experimental.enable_op_determinism()
    except Exception:
        pass


# --------------------------------------------------
# Residual causal TCN block
# --------------------------------------------------

def residual_tcn_block(
    x,
    filters,
    dilation_rate,
    dropout_rate=0.20
):

    residual = x

    x = layers.Conv1D(
        filters=filters,
        kernel_size=3,
        padding="causal",
        dilation_rate=dilation_rate
    )(x)

    x = layers.BatchNormalization()(x)

    x = layers.ReLU()(x)

    x = layers.Dropout(
        dropout_rate
    )(x)

    if residual.shape[-1] != filters:

        residual = layers.Conv1D(
            filters=filters,
            kernel_size=1,
            padding="same"
        )(residual)

    x = layers.Add()(
        [x, residual]
    )

    x = layers.ReLU()(x)

    return x


# --------------------------------------------------
# GRIDGUARD TCN
# --------------------------------------------------

def build_gridguard_tcn(
    input_shape=(20, 141)
):

    inputs = layers.Input(
        shape=input_shape,
        name="grid_telemetry"
    )

    x = residual_tcn_block(
        inputs,
        filters=32,
        dilation_rate=1
    )

    x = residual_tcn_block(
        x,
        filters=32,
        dilation_rate=2
    )

    x = residual_tcn_block(
        x,
        filters=64,
        dilation_rate=4
    )

    x = layers.GlobalAveragePooling1D()(x)

    x = layers.Dense(
        32,
        activation="relu"
    )(x)

    x = layers.Dropout(
        0.30
    )(x)

    outputs = layers.Dense(
        1,
        activation="sigmoid",
        name="attack_probability"
    )(x)

    model = Model(
        inputs=inputs,
        outputs=outputs,
        name="GRIDGUARD_TCN"
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=LEARNING_RATE
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


# --------------------------------------------------
# Callback builder
# --------------------------------------------------

def make_tcn_callbacks():

    return [

        tf.keras.callbacks.EarlyStopping(
            monitor="val_loss",
            patience=EARLY_STOP_PATIENCE,
            restore_best_weights=True
        ),

        tf.keras.callbacks.ReduceLROnPlateau(
            monitor="val_loss",
            factor=LR_REDUCE_FACTOR,
            patience=LR_REDUCE_PATIENCE,
            min_lr=MIN_LR
        )
    ]


# --------------------------------------------------
# Verification only
# --------------------------------------------------

set_global_seed(11)

model_check = build_gridguard_tcn()

print("Seeds:", SEEDS)
print("Batch size:", BATCH_SIZE)
print("Max epochs:", MAX_EPOCHS)
print("Learning rate:", LEARNING_RATE)
print("Early-stop patience:", EARLY_STOP_PATIENCE)

print("\nInput shape :", model_check.input_shape)
print("Output shape:", model_check.output_shape)
print("Parameters  :", model_check.count_params())
```

```python

# ============================================================
# STEP 32 — GRIDGUARD-AI STRICT AUTOENCODER LOSO
# 8 held-out scenarios | Benign-only training
# Validation-calibrated anomaly threshold | No test leakage
# ============================================================

import os
import gc
import random
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix
from tensorflow.keras import layers, Model

OUTPUT_DIR = "/kaggle/working/gridguard_etep_final"
os.makedirs(OUTPUT_DIR, exist_ok=True)

AE_SEED = 42
AE_EPOCHS = 25
AE_BATCH = 64
AE_TARGET_VAL_FAR = 0.05

# Reuse the strict features, labels, scenarios and
# purged chronological LOSO splitter from previous cells.
required = [
    "X_seq_strict", "y_seq_strict", "scenario_seq",
    "scenario_folders", "make_tcn_fold_indices"
]
missing = [name for name in required if name not in globals()]
if missing:
    raise RuntimeError(
        f"Run earlier preprocessing/split cells first: {missing}"
    )

X_all = np.asarray(X_seq_strict)
y_all = np.asarray(y_seq_strict).astype(int)

assert X_all.ndim == 3
assert X_all.shape[0] == len(y_all) == len(scenario_seq)
assert set(np.unique(y_all)).issubset({0, 1})

print("Dataset shape:", X_all.shape)
print("Scenarios:", len(scenario_folders))

def build_ae(n_features):
    inputs = layers.Input(shape=(n_features,))
    x = layers.Dense(96, activation="relu")(inputs)
    x = layers.Dense(48, activation="relu")(x)
    x = layers.Dense(16, activation="relu")(x)
    x = layers.Dense(48, activation="relu")(x)
    x = layers.Dense(96, activation="relu")(x)
    outputs = layers.Dense(n_features)(x)

    model = Model(inputs, outputs)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(1e-3),
        loss="mse"
    )
    return model

def reconstruction_error(model, X):
    predictions = model.predict(
        X, batch_size=256, verbose=0
    )
    return np.mean((X - predictions) ** 2, axis=1)

def evaluate(y_true, y_pred):
    tn, fp, fn, tp = confusion_matrix(
        y_true, y_pred, labels=[0, 1]
    ).ravel()

    return {
        "TP": int(tp),
        "FN": int(fn),
        "FP": int(fp),
        "TN": int(tn),
        "attack_recall": tp / (tp + fn) if tp + fn else np.nan,
        "FSR": fn / (tp + fn) if tp + fn else np.nan,
        "benign_FAR": fp / (fp + tn) if fp + tn else np.nan,
    }

results = []
prediction_rows = []

for fold_number, held_out in enumerate(scenario_folders, start=1):

    print(f"\n{'='*65}")
    print(f"AE FOLD {fold_number}/{len(scenario_folders)}")
    print("Held-out scenario:", held_out)

    fit_idx, val_idx, test_idx = make_tcn_fold_indices(
        held_out
    )

    # Check strict scenario isolation
    assert not np.isin(fit_idx, test_idx).any()
    assert not np.isin(val_idx, test_idx).any()
    assert np.all(
        np.asarray(scenario_seq)[test_idx] == held_out
    )

    # Fit only on BENIGN windows
    benign_fit_idx = fit_idx[y_all[fit_idx] == 0]
    benign_val_idx = val_idx[y_all[val_idx] == 0]

    if len(benign_fit_idx) < 100 or len(benign_val_idx) < 20:
        raise RuntimeError(
            f"Insufficient benign samples for fold {held_out}"
        )

    # Last timestep as window-level feature representation.
    # Do not mix this AE with a sequence-reconstruction AE.
    X_fit = X_all[benign_fit_idx, -1, :]
    X_val = X_all[benign_val_idx, -1, :]
    X_test = X_all[test_idx, -1, :]

    # Scaling uses benign FIT data only
    scaler = StandardScaler()
    X_fit = scaler.fit_transform(X_fit).astype(np.float32)
    X_val = scaler.transform(X_val).astype(np.float32)
    X_test = scaler.transform(X_test).astype(np.float32)

    assert np.isfinite(X_fit).all()
    assert np.isfinite(X_val).all()
    assert np.isfinite(X_test).all()

    tf.keras.backend.clear_session()
    random.seed(AE_SEED)
    np.random.seed(AE_SEED)
    tf.keras.utils.set_random_seed(AE_SEED)

    model = build_ae(X_fit.shape[1])

    history = model.fit(
        X_fit, X_fit,
        validation_data=(X_val, X_val),
        epochs=AE_EPOCHS,
        batch_size=AE_BATCH,
        shuffle=True,
        verbose=0,
        callbacks=[
            tf.keras.callbacks.EarlyStopping(
                monitor="val_loss",
                patience=4,
                restore_best_weights=True
            )
        ]
    )

    # The anomaly threshold is chosen using
    # benign validation reconstruction errors ONLY.
    val_errors = reconstruction_error(model, X_val)
    threshold = float(
        np.quantile(val_errors, 1.0 - AE_TARGET_VAL_FAR)
    )

    test_errors = reconstruction_error(model, X_test)
    test_pred = (test_errors > threshold).astype(int)

    y_test = y_all[test_idx]
    metrics = evaluate(y_test, test_pred)

    results.append({
        "scenario": held_out,
        "n_test": len(test_idx),
        "n_benign_fit": len(benign_fit_idx),
        "n_benign_validation": len(benign_val_idx),
        "threshold": threshold,
        "epochs": len(history.history["loss"]),
        **metrics
    })

    for idx, true_label, score, pred in zip(
        test_idx, y_test, test_errors, test_pred
    ):
        prediction_rows.append({
            "scenario": held_out,
            "global_index": int(idx),
            "y_true": int(true_label),
            "ae_error": float(score),
            "ae_pred": int(pred),
            "ae_threshold": threshold
        })

    # Save checkpoints after every fold
    pd.DataFrame(results).to_csv(
        f"{OUTPUT_DIR}/ae_loso_results.csv", index=False
    )
    pd.DataFrame(prediction_rows).to_csv(
        f"{OUTPUT_DIR}/ae_loso_predictions.csv", index=False
    )

    print(
        f"Attack recall: {metrics['attack_recall']:.4f} | "
        f"FSR: {metrics['FSR']:.4f} | "
        f"Benign FAR: {metrics['benign_FAR']:.4f}"
    )

    del model, scaler, X_fit, X_val, X_test
    gc.collect()

ae_results = pd.DataFrame(results)
ae_predictions = pd.DataFrame(prediction_rows)

assert len(ae_results) == len(scenario_folders)
assert len(ae_predictions) == len(y_all)
assert not ae_predictions.global_index.duplicated().any()

print("\n========== FINAL AE LOSO RESULTS ==========")
print(ae_results.to_string(index=False))

print("\nMacro averages:")
print(
    ae_results[["attack_recall", "FSR", "benign_FAR"]]
    .mean()
)

print("\nSaved files:")
print(f"{OUTPUT_DIR}/ae_loso_results.csv")
print(f"{OUTPUT_DIR}/ae_loso_predictions.csv")
```

```python

# ============================================================
# STEP 32 — GRIDGUARD-AI AUTOENCODER LOSO
# Benign-only training | Validation-only threshold
# Held-out scenario evaluation
# ============================================================

import gc
from pathlib import Path
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix
from tensorflow.keras import layers, Model

OUT = Path("/kaggle/working/gridguard_etep_final")
OUT.mkdir(parents=True, exist_ok=True)

required = [
    "X_seq_strict", "y_seq_strict", "scenario_seq",
    "scenario_folders", "make_tcn_fold_indices"
]
missing = [n for n in required if n not in globals()]
assert not missing, f"Missing variables: {missing}"

X_all = np.asarray(X_seq_strict)
y_all = np.asarray(y_seq_strict).astype(int)
s_all = np.asarray(scenario_seq)

assert X_all.shape == (14202, 20, 141)
assert len(y_all) == len(s_all) == len(X_all)

def build_autoencoder(n_features):
    inp = layers.Input(shape=(n_features,))
    x = layers.Dense(96, activation="relu")(inp)
    x = layers.Dense(48, activation="relu")(x)
    x = layers.Dense(16, activation="relu")(x)
    x = layers.Dense(48, activation="relu")(x)
    x = layers.Dense(96, activation="relu")(x)
    out = layers.Dense(n_features)(x)

    model = Model(inp, out)
    model.compile(optimizer="adam", loss="mse")
    return model

def reconstruction_error(model, x):
    reconstructed = model.predict(x, batch_size=256, verbose=0)
    return np.mean((x - reconstructed) ** 2, axis=1)

results = []
predictions = []

for k, held_out in enumerate(scenario_folders, start=1):
    print(f"\n========== AE FOLD {k}/8: {held_out} ==========", flush=True)

    fit_idx, val_idx, test_idx = make_tcn_fold_indices(held_out)

    benign_fit = fit_idx[y_all[fit_idx] == 0]
    benign_val = val_idx[y_all[val_idx] == 0]

    if len(benign_fit) < 100 or len(benign_val) < 20:
        raise RuntimeError(
            f"Insufficient benign data in {held_out}"
        )

    # Window representation: final timestep only
    X_fit = X_all[benign_fit, -1, :]
    X_val = X_all[benign_val, -1, :]
    X_test = X_all[test_idx, -1, :]

    scaler = StandardScaler()
    X_fit = scaler.fit_transform(X_fit).astype("float32")
    X_val = scaler.transform(X_val).astype("float32")
    X_test = scaler.transform(X_test).astype("float32")

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(42)

    model = build_autoencoder(X_fit.shape[1])

    history = model.fit(
        X_fit, X_fit,
        validation_data=(X_val, X_val),
        epochs=25,
        batch_size=64,
        shuffle=True,
        verbose=0,
        callbacks=[
            tf.keras.callbacks.EarlyStopping(
                monitor="val_loss",
                patience=4,
                restore_best_weights=True
            )
        ]
    )

    val_error = reconstruction_error(model, X_val)
    threshold = float(np.quantile(val_error, 0.95))

    test_error = reconstruction_error(model, X_test)
    y_test = y_all[test_idx]
    y_pred = (test_error > threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_test, y_pred, labels=[0, 1]
    ).ravel()

    recall = tp / (tp + fn) if tp + fn else np.nan
    fsr = fn / (tp + fn) if tp + fn else np.nan
    far = fp / (fp + tn) if fp + tn else np.nan

    results.append({
        "scenario": held_out,
        "n_test": len(test_idx),
        "threshold": threshold,
        "attack_recall": recall,
        "FSR": fsr,
        "benign_FAR": far,
        "TP": tp, "FN": fn, "FP": fp, "TN": tn,
        "epochs": len(history.history["loss"])
    })

    for idx, label, score, pred in zip(
        test_idx, y_test, test_error, y_pred
    ):
        predictions.append({
            "global_index": int(idx),
            "scenario": held_out,
            "y_true": int(label),
            "ae_error": float(score),
            "ae_pred": int(pred),
            "ae_threshold": threshold
        })

    pd.DataFrame(results).to_csv(
        OUT / "ae_loso_results.csv", index=False
    )
    pd.DataFrame(predictions).to_csv(
        OUT / "ae_loso_predictions.csv", index=False
    )

    print(
        f"Recall={recall:.4f} | "
        f"FSR={fsr:.4f} | FAR={far:.4f} | "
        f"Epochs={len(history.history['loss'])}",
        flush=True
    )

    del model, X_fit, X_val, X_test
    gc.collect()

ae_results = pd.DataFrame(results)
ae_predictions = pd.DataFrame(predictions)

assert len(ae_results) == 8
assert len(ae_predictions) == len(X_all)
assert ae_predictions.global_index.is_unique

print("\n========== FINAL AE LOSO RESULTS ==========")
print(ae_results.to_string(index=False))

print("\nMacro averages:")
print(
    ae_results[
        ["attack_recall", "FSR", "benign_FAR"]
    ].mean().to_string()
)

print("\nSAVED:")
print(OUT / "ae_loso_results.csv")
print(OUT / "ae_loso_predictions.csv")
```

```python

# STEP 38 — RESTORE ORIGINAL TCN FUNCTIONS
# No training in this cell

import os, random
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
from sklearn.preprocessing import StandardScaler

SEEDS = [11, 42, 77]
BATCH_SIZE = 64
MAX_EPOCHS = 25
LEARNING_RATE = 1e-3
EARLY_STOP_PATIENCE = 4

def set_global_seed(seed):
    os.environ["PYTHONHASHSEED"] = str(seed)
    random.seed(seed)
    np.random.seed(seed)
    tf.keras.utils.set_random_seed(seed)
    try:
        tf.config.experimental.enable_op_determinism()
    except Exception:
        pass

def prepare_tcn_fold(held_out_scenario):
    fit_idx, val_idx, test_idx = make_tcn_fold_indices(
        held_out_scenario
    )

    X_fit = np.asarray(X_seq_strict[fit_idx]).copy()
    X_val = np.asarray(X_seq_strict[val_idx]).copy()
    X_test = np.asarray(X_seq_strict[test_idx]).copy()

    y_fit = np.asarray(y_seq_strict[fit_idx]).copy()
    y_val = np.asarray(y_seq_strict[val_idx]).copy()
    y_test = np.asarray(y_seq_strict[test_idx]).copy()

    nf = X_fit.shape[-1]
    scaler = StandardScaler()
    scaler.fit(X_fit.reshape(-1, nf))

    def transform(x):
        return scaler.transform(
            x.reshape(-1, nf)
        ).reshape(x.shape).astype(np.float32)

    return (
        transform(X_fit), y_fit,
        transform(X_val), y_val,
        transform(X_test), y_test,
        scaler
    )

def residual_tcn_block(x, filters, dilation_rate,
                       dropout_rate=0.20):
    residual = x

    x = layers.Conv1D(
        filters, kernel_size=3,
        padding="causal",
        dilation_rate=dilation_rate
    )(x)
    x = layers.BatchNormalization()(x)
    x = layers.ReLU()(x)
    x = layers.Dropout(dropout_rate)(x)

    if residual.shape[-1] != filters:
        residual = layers.Conv1D(
            filters, kernel_size=1, padding="same"
        )(residual)

    x = layers.Add()([x, residual])
    return layers.ReLU()(x)

def build_gridguard_tcn(input_shape=(20, 141)):
    inputs = layers.Input(
        shape=input_shape, name="grid_telemetry"
    )

    x = residual_tcn_block(inputs, 32, 1)
    x = residual_tcn_block(x, 32, 2)
    x = residual_tcn_block(x, 64, 4)

    x = layers.GlobalAveragePooling1D()(x)
    x = layers.Dense(32, activation="relu")(x)
    x = layers.Dropout(0.30)(x)

    outputs = layers.Dense(
        1, activation="sigmoid",
        name="attack_probability"
    )(x)

    model = Model(inputs, outputs, name="GRIDGUARD_TCN")
    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=LEARNING_RATE
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )
    return model

def make_tcn_callbacks():
    return [
        tf.keras.callbacks.EarlyStopping(
            monitor="val_loss",
            patience=EARLY_STOP_PATIENCE,
            restore_best_weights=True
        )
    ]

required = [
    "X_seq_strict", "y_seq_strict",
    "scenario_seq", "scenario_folders",
    "make_tcn_fold_indices",
    "prepare_tcn_fold", "build_gridguard_tcn",
    "set_global_seed", "make_tcn_callbacks", "SEEDS"
]

missing = [name for name in required if name not in globals()]
assert not missing, f"Still missing: {missing}"

assert np.asarray(X_seq_strict).shape == (14202, 20, 141)

set_global_seed(11)
model_check = build_gridguard_tcn()

print("TCN READY")
print("Seeds:", SEEDS)
print("Epochs:", MAX_EPOCHS)
print("Parameters:", model_check.count_params())
print("Ready for 8 folds x 3 seeds = 24 runs")
```

```python

# STEP 39 — TCN LOSO TRAINING + PREDICTION EXPORT
# 8 scenarios x 3 seeds = 24 runs

import gc
import numpy as np
import pandas as pd
import tensorflow as tf
from pathlib import Path
from sklearn.metrics import confusion_matrix

OUT = Path("/kaggle/working/gridguard_etep_final")
OUT.mkdir(parents=True, exist_ok=True)

test_file = OUT / "gridguard_tcn_test_predictions_long.csv"
val_file = OUT / "gridguard_tcn_validation_predictions_long.csv"
result_file = OUT / "gridguard_tcn_seed_results.csv"

# Restart-safe: skip fully saved runs
if result_file.exists():
    results = pd.read_csv(result_file).to_dict("records")
else:
    results = []

test_rows = (
    pd.read_csv(test_file).to_dict("records")
    if test_file.exists() else []
)
val_rows = (
    pd.read_csv(val_file).to_dict("records")
    if val_file.exists() else []
)

completed = {
    (str(r["scenario"]), int(r["seed"]))
    for r in results
}

for fold_no, scenario in enumerate(scenario_folders, 1):

    fit_idx, val_idx, test_idx = make_tcn_fold_indices(scenario)

    X_fit, y_fit, X_val, y_val, X_test, y_test, scaler = (
        prepare_tcn_fold(scenario)
    )

    for seed in SEEDS:

        if (scenario, seed) in completed:
            print("SKIPPING SAVED:", scenario, seed)
            continue

        print(
            f"\nFOLD {fold_no}/8 | SEED {seed}",
            flush=True
        )

        tf.keras.backend.clear_session()
        set_global_seed(seed)

        model = build_gridguard_tcn()

        history = model.fit(
            X_fit, y_fit,
            validation_data=(X_val, y_val),
            epochs=MAX_EPOCHS,
            batch_size=BATCH_SIZE,
            callbacks=make_tcn_callbacks(),
            verbose=0
        )

        p_test = model.predict(
            X_test, batch_size=256, verbose=0
        ).ravel()

        p_val = model.predict(
            X_val, batch_size=256, verbose=0
        ).ravel()

        pred = (p_test >= 0.5).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_test, pred, labels=[0, 1]
        ).ravel()

        recall = tp / (tp + fn) if tp + fn else np.nan
        far = fp / (fp + tn) if fp + tn else np.nan

        test_rows.extend({
            "global_index": int(idx),
            "scenario": scenario,
            "seed": int(seed),
            "y_true": int(label),
            "tcn_probability": float(prob)
        } for idx, label, prob in zip(
            test_idx, y_test, p_test
        ))

        val_rows.extend({
            "held_out_scenario": scenario,
            "global_index": int(idx),
            "seed": int(seed),
            "y_true": int(label),
            "tcn_probability": float(prob)
        } for idx, label, prob in zip(
            val_idx, y_val, p_val
        ))

        results.append({
            "scenario": scenario,
            "seed": seed,
            "epochs": len(history.history["loss"]),
            "TP": int(tp),
            "FN": int(fn),
            "FP": int(fp),
            "TN": int(tn),
            "attack_recall": recall,
            "benign_FAR": far
        })

        # Save immediately after each successful run
        pd.DataFrame(test_rows).to_csv(test_file, index=False)
        pd.DataFrame(val_rows).to_csv(val_file, index=False)
        pd.DataFrame(results).to_csv(result_file, index=False)

        print(
            f"Recall={recall:.4f} | FAR={far:.4f}",
            flush=True
        )

        del model
        gc.collect()

print("\n=== ALL TCN RUNS COMPLETE ===")
print("Runs:", len(results))
print("Test prediction rows:", len(test_rows))
print("Validation prediction rows:", len(val_rows))

assert len(results) == 24
assert len(test_rows) == len(y_seq_strict) * 3
assert not pd.DataFrame(test_rows).duplicated(
    ["global_index", "seed"]
).any()

print("\nFILES SAVED:")
print(test_file)
print(val_file)
print(result_file)
```

```python

# STEP 40 — RESUME TCN TRAINING
# Preserve completed runs and save after every new run

import gc
import numpy as np
import pandas as pd
import tensorflow as tf
from pathlib import Path
from sklearn.metrics import confusion_matrix

OUT = Path("/kaggle/working/gridguard_etep_final")

results_path = OUT / "gridguard_tcn_seed_results.csv"
test_path = OUT / "gridguard_tcn_test_predictions_long.csv"
val_path = OUT / "gridguard_tcn_validation_predictions_long.csv"

results = pd.read_csv(results_path)
test_df = pd.read_csv(test_path)
val_df = pd.read_csv(val_path)

# Check all three checkpoints before skipping any run
required_results = {"scenario", "seed"}
required_test = {"scenario", "seed", "global_index", "y_true", "tcn_probability"}
required_val = {"held_out_scenario", "seed", "global_index", "y_true", "tcn_probability"}

assert required_results.issubset(results.columns)
assert required_test.issubset(test_df.columns)
assert required_val.issubset(val_df.columns)

completed = set()

for scenario in scenario_folders:
    fit_idx, val_idx, test_idx = make_tcn_fold_indices(scenario)

    for seed in SEEDS:
        has_result = (
            (results["scenario"] == scenario) &
            (results["seed"] == seed)
        ).sum() == 1

        n_test = (
            (test_df["scenario"] == scenario) &
            (test_df["seed"] == seed)
        ).sum()

        n_val = (
            (val_df["held_out_scenario"] == scenario) &
            (val_df["seed"] == seed)
        ).sum()

        if has_result and n_test == len(test_idx) and n_val == len(val_idx):
            completed.add((scenario, seed))
        elif has_result or n_test or n_val:
            raise RuntimeError(
                f"Incomplete checkpoint for {scenario}, seed {seed}. "
                "Do not overwrite; inspect saved CSVs first."
            )

print("Verified completed:", len(completed))
print("Remaining:", 24 - len(completed))

for fold_number, scenario in enumerate(scenario_folders, 1):

    pending = [
        seed for seed in SEEDS
        if (scenario, seed) not in completed
    ]

    if not pending:
        continue

    fit_idx, val_idx, test_idx = make_tcn_fold_indices(scenario)

    X_fit, y_fit, X_val, y_val, X_test, y_test, scaler = (
        prepare_tcn_fold(scenario)
    )

    for seed in pending:
        print(
            f"\nTRAINING FOLD {fold_number}/8 — SEED {seed}",
            flush=True
        )

        tf.keras.backend.clear_session()
        set_global_seed(seed)

        model = build_gridguard_tcn()

        history = model.fit(
            X_fit, y_fit,
            validation_data=(X_val, y_val),
            epochs=MAX_EPOCHS,
            batch_size=BATCH_SIZE,
            callbacks=make_tcn_callbacks(),
            verbose=0
        )

        p_test = model.predict(
            X_test, batch_size=256, verbose=0
        ).reshape(-1)

        p_val = model.predict(
            X_val, batch_size=256, verbose=0
        ).reshape(-1)

        y_pred = (p_test >= 0.5).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_test, y_pred, labels=[0, 1]
        ).ravel()

        recall = tp / (tp + fn) if tp + fn else np.nan
        far = fp / (fp + tn) if fp + tn else np.nan

        new_test = pd.DataFrame({
            "global_index": test_idx,
            "scenario": scenario,
            "seed": seed,
            "y_true": y_test,
            "tcn_probability": p_test
        })

        new_val = pd.DataFrame({
            "held_out_scenario": scenario,
            "global_index": val_idx,
            "seed": seed,
            "y_true": y_val,
            "tcn_probability": p_val
        })

        new_result = pd.DataFrame([{
            "scenario": scenario,
            "seed": seed,
            "epochs": len(history.history["loss"]),
            "TP": int(tp),
            "FN": int(fn),
            "FP": int(fp),
            "TN": int(tn),
            "attack_recall": recall,
            "benign_FAR": far
        }])

        # Write prediction files first; results file last
        updated_test = pd.concat(
            [test_df, new_test], ignore_index=True
        )
        updated_val = pd.concat(
            [val_df, new_val], ignore_index=True
        )
        updated_results = pd.concat(
            [results, new_result], ignore_index=True
        )

        updated_test.to_csv(test_path, index=False)
        updated_val.to_csv(val_path, index=False)
        updated_results.to_csv(results_path, index=False)

        test_df = updated_test
        val_df = updated_val
        results = updated_results

        completed.add((scenario, seed))

        print(
            f"SAVED {len(completed)}/24 | "
            f"Recall={recall:.4f} | FAR={far:.4f}",
            flush=True
        )

        del model
        gc.collect()

    del X_fit, X_val, X_test, scaler
    gc.collect()

print("\nTCN TRAINING COMPLETE")
print("Runs:", len(results))
print("Test prediction rows:", len(test_df))

assert len(results) == 24
assert len(test_df) == 42606
assert not test_df.duplicated(["global_index", "seed"]).any()

print("ALL 24 TCN RUNS SAVED")
```

```python

# STEP 42 — Leakage-safe RF LOSO
# Preserves existing RF CSV; trains on FIT partition only

import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix

OUT = Path("/kaggle/working/gridguard_etep_final")

X = np.asarray(X_seq_strict)[:, -1, :]
y = np.asarray(y_seq_strict).astype(int)
s = np.asarray(scenario_seq)

test_rows = []
val_rows = []
metrics = []

for fold in scenario_folders:
    fit_idx, val_idx, test_idx = make_tcn_fold_indices(fold)

    model = RandomForestClassifier(
        n_estimators=120,
        max_depth=18,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )

    model.fit(X[fit_idx], y[fit_idx])

    # Works even if training contains only one class
    classes = model.classes_
    attack_col = np.where(classes == 1)[0]

    if len(attack_col) == 1:
        p_test = model.predict_proba(X[test_idx])[:, attack_col[0]]
        p_val = model.predict_proba(X[val_idx])[:, attack_col[0]]
    else:
        p_test = np.full(len(test_idx), float(classes[0]))
        p_val = np.full(len(val_idx), float(classes[0]))

    pred = (p_test >= 0.5).astype(int)
    tn, fp, fn, tp = confusion_matrix(
        y[test_idx], pred, labels=[0, 1]
    ).ravel()

    test_rows.append(pd.DataFrame({
        "global_index": test_idx,
        "scenario": s[test_idx],
        "y_true": y[test_idx],
        "rf_probability": p_test
    }))

    val_rows.append(pd.DataFrame({
        "held_out_scenario": fold,
        "global_index": val_idx,
        "scenario": s[val_idx],
        "y_true": y[val_idx],
        "rf_probability": p_val
    }))

    metrics.append({
        "scenario": fold,
        "TP": tp, "FN": fn, "FP": fp, "TN": tn,
        "attack_recall": tp / (tp + fn) if tp + fn else np.nan,
        "benign_FAR": fp / (fp + tn) if fp + tn else np.nan
    })

    print("RF COMPLETE:", fold, flush=True)

rf_test = pd.concat(test_rows, ignore_index=True)
rf_val = pd.concat(val_rows, ignore_index=True)

assert len(rf_test) == 14202
assert rf_test.global_index.is_unique

rf_test.to_csv(
    OUT / "gridguard_rf_loso_predictions_strict.csv",
    index=False
)
rf_val.to_csv(
    OUT / "gridguard_rf_validation_predictions_strict.csv",
    index=False
)
pd.DataFrame(metrics).to_csv(
    OUT / "gridguard_rf_loso_results_strict.csv",
    index=False
)

print("\nSTRICT RF COMPLETE")
print("Test predictions:", len(rf_test))
print("Validation predictions:", len(rf_val))
print(pd.DataFrame(metrics).to_string(index=False))
```

```python

# STEP 43 — VALIDATION-CALIBRATED ASSURANCE FUSION
# Uses STRICT RF + TCN + AE test predictions

from pathlib import Path
import numpy as np
import pandas as pd

P = Path("/kaggle/working/gridguard_etep_final")

rf = pd.read_csv(P / "gridguard_rf_loso_predictions_strict.csv")
rfv = pd.read_csv(P / "gridguard_rf_validation_predictions_strict.csv")
tc = pd.read_csv(P / "gridguard_tcn_test_predictions_long.csv")
tv = pd.read_csv(P / "gridguard_tcn_validation_predictions_long.csv")
ae = pd.read_csv(P / "ae_loso_predictions.csv")

keys = ["global_index", "scenario", "y_true"]

assert rf.global_index.is_unique
assert ae.global_index.is_unique

# Aggregate exactly 3 independent TCN seed predictions
assert not tc.duplicated(["global_index", "seed"]).any()

tc3 = tc.groupby(keys, as_index=False).agg(
    tcn_probability=("tcn_probability", "mean"),
    seed_count=("seed", "nunique")
)
assert (tc3.seed_count == 3).all()

test = rf.merge(tc3, on=keys, validate="one_to_one")
test = test.merge(ae, on=keys, validate="one_to_one")

assert len(test) == 14202
assert test.global_index.is_unique

# Validation is fold-specific; do not pool across held-out folds
assert not tv.duplicated(
    ["held_out_scenario", "global_index", "seed"]
).any()

tv3 = tv.groupby(
    ["held_out_scenario", "global_index", "y_true"],
    as_index=False
).agg(
    tcn_probability=("tcn_probability", "mean"),
    seed_count=("seed", "nunique")
)

assert (tv3.seed_count == 3).all()

validation = rfv.merge(
    tv3,
    on=["held_out_scenario", "global_index", "y_true"],
    validate="one_to_one"
)

assert len(validation) == len(rfv)

outputs = []
thresholds = []

for scenario in scenario_folders:

    v = validation[
        validation.held_out_scenario == scenario
    ]

    t = test[test.scenario == scenario].copy()

    benign = v[v.y_true == 0]

    if len(benign) < 20:
        raise RuntimeError(
            f"Insufficient benign validation: {scenario}"
        )

    # Thresholds selected WITHOUT viewing test labels
    rf_low = benign.rf_probability.quantile(0.25)
    tcn_low = benign.tcn_probability.quantile(0.25)

    rf_high = benign.rf_probability.quantile(0.99)
    tcn_high = benign.tcn_probability.quantile(0.99)

    # AUTO ALERT only when BOTH detectors strongly agree
    alert = (
        (t.rf_probability > rf_high) &
        (t.tcn_probability > tcn_high)
    )

    # SAFE requires BOTH low risk and AE not anomalous
    safe = (
        (t.rf_probability < rf_low) &
        (t.tcn_probability < tcn_low) &
        (t.ae_pred == 0)
    )

    t["decision"] = np.select(
        [alert, safe],
        ["AUTO ALERT", "SAFE"],
        default="HUMAN REVIEW"
    )

    outputs.append(t)

    thresholds.append({
        "scenario": scenario,
        "rf_low": rf_low,
        "tcn_low": tcn_low,
        "rf_high": rf_high,
        "tcn_high": tcn_high,
        "n_benign_validation": len(benign)
    })

fusion = pd.concat(outputs, ignore_index=True)

report = []

for scenario, g in fusion.groupby("scenario", sort=False):

    attack = g[g.y_true == 1]
    benign = g[g.y_true == 0]

    report.append({
        "scenario": scenario,
        "n": len(g),
        "attack_safe_rate": (
            (attack.decision == "SAFE").mean()
            if len(attack) else np.nan
        ),
        "attack_escalation_rate": (
            (attack.decision != "SAFE").mean()
            if len(attack) else np.nan
        ),
        "benign_safe_rate": (
            (benign.decision == "SAFE").mean()
            if len(benign) else np.nan
        ),
        "human_review_rate": (
            (g.decision == "HUMAN REVIEW").mean()
        ),
        "auto_alert_rate": (
            (g.decision == "AUTO ALERT").mean()
        )
    })

report = pd.DataFrame(report)

fusion.to_csv(P / "gridguard_fusion_decisions.csv", index=False)
report.to_csv(P / "gridguard_fusion_results.csv", index=False)

pd.DataFrame(thresholds).to_csv(
    P / "gridguard_fusion_thresholds.csv", index=False
)

print("FUSION COMPLETE")
print("Evaluated windows:", len(fusion))
print("\nDecision counts:")
print(fusion.decision.value_counts().to_string())

print("\nSCENARIO RESULTS:")
print(report.to_string(index=False))

print("\nSaved all fusion CSV files.")
```

```python

# GRIDGUARD-AI — FINAL EVALUATION + ASSURANCE + HITL
# No retraining | Real predictions | CSV outputs

from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, confusion_matrix

P = Path("/kaggle/working/gridguard_etep_final")
P.mkdir(parents=True, exist_ok=True)

rf = pd.read_csv(P / "gridguard_rf_loso_predictions_strict.csv")
tcn = pd.read_csv(P / "gridguard_tcn_test_predictions_long.csv")
ae = pd.read_csv(P / "ae_loso_predictions.csv")
rv = pd.read_csv(P / "gridguard_rf_validation_predictions_strict.csv")
tv = pd.read_csv(P / "gridguard_tcn_validation_predictions_long.csv")

KEY = ["global_index", "scenario", "y_true"]

# 1 — Merge three TCN seeds
assert not tcn.duplicated(["global_index", "seed"]).any()

tcn_mean = tcn.groupby(KEY, as_index=False).agg(
    tcn_probability=("tcn_probability", "mean"),
    seeds=("seed", "nunique")
)
assert (tcn_mean.seeds == 3).all()

df = rf.merge(tcn_mean, on=KEY, validate="one_to_one")
df = df.merge(ae, on=KEY, validate="one_to_one")

assert len(df) == 14202
assert df.global_index.is_unique

# 2 — Fold-specific validation data
tv_mean = tv.groupby(
    ["held_out_scenario", "global_index", "y_true"],
    as_index=False
)["tcn_probability"].mean()

validation = rv.merge(
    tv_mean,
    on=["held_out_scenario", "global_index", "y_true"],
    validate="one_to_one"
)

# 3 — Fixed assurance policy calibrated on benign validation
# Post-hoc exploratory evaluation; test outcomes already inspected.
reports = []
decisions = []

for scenario in df.scenario.unique():
    test_fold = df[df.scenario == scenario].copy()
    val_fold = validation[
        validation.held_out_scenario == scenario
    ]

    benign_val = val_fold[val_fold.y_true == 0]
    assert len(benign_val) >= 20

    # Predefined conservative quantiles
    tcn_low = benign_val.tcn_probability.quantile(0.05)
    rf_low = benign_val.rf_probability.quantile(0.05)
    tcn_high = benign_val.tcn_probability.quantile(0.95)

    # Alert requires strong TCN evidence + independent AE flag
    alert = (
        (test_fold.tcn_probability >= tcn_high) &
        (test_fold.ae_pred == 1)
    )

    # Safe requires agreement of all three indicators
    safe = (
        (test_fold.tcn_probability < tcn_low) &
        (test_fold.rf_probability < rf_low) &
        (test_fold.ae_pred == 0)
    )

    test_fold["decision"] = np.select(
        [alert, safe],
        ["AUTO ALERT", "SAFE"],
        default="HUMAN REVIEW"
    )

    test_fold["requires_human"] = (
        test_fold.decision == "HUMAN REVIEW"
    ).astype(int)

    test_fold["evidence_summary"] = (
        "TCN=" + test_fold.tcn_probability.round(4).astype(str) +
        "; RF=" + test_fold.rf_probability.round(4).astype(str) +
        "; AE_flag=" + test_fold.ae_pred.astype(str)
    )

    decisions.append(test_fold)

    y = test_fold.y_true.astype(int)
    attack = test_fold[y == 1]
    benign = test_fold[y == 0]

    reports.append({
        "scenario": scenario,
        "n": len(test_fold),
        "rf_auc": (
            roc_auc_score(y, test_fold.rf_probability)
            if y.nunique() == 2 else np.nan
        ),
        "tcn_auc": (
            roc_auc_score(y, test_fold.tcn_probability)
            if y.nunique() == 2 else np.nan
        ),
        "attack_false_safe_rate": (
            (attack.decision == "SAFE").mean()
            if len(attack) else np.nan
        ),
        "attack_escalation_rate": (
            (attack.decision != "SAFE").mean()
            if len(attack) else np.nan
        ),
        "benign_safe_release_rate": (
            (benign.decision == "SAFE").mean()
            if len(benign) else np.nan
        ),
        "human_review_rate": (
            test_fold.requires_human.mean()
        ),
        "auto_alert_rate": (
            (test_fold.decision == "AUTO ALERT").mean()
        )
    })

final = pd.concat(decisions, ignore_index=True)
report = pd.DataFrame(reports)

# 4 — HITL simulation: humans receive review requests.
# No actual human decisions are simulated or fabricated.
human_queue = final[
    final.decision == "HUMAN REVIEW"
].copy()

human_queue["human_status"] = "PENDING_REVIEW"

# 5 — Safety audit
attacks = final[final.y_true == 1]
benign = final[final.y_true == 0]

summary = pd.DataFrame([{
    "total_windows": len(final),
    "attack_windows": len(attacks),
    "benign_windows": len(benign),
    "unsafe_safe_releases": int(
        (attacks.decision == "SAFE").sum()
    ),
    "attack_false_safe_rate": (
        (attacks.decision == "SAFE").mean()
    ),
    "attack_escalation_rate": (
        (attacks.decision != "SAFE").mean()
    ),
    "benign_safe_release_rate": (
        (benign.decision == "SAFE").mean()
    ),
    "human_review_count": len(human_queue),
    "human_review_rate": len(human_queue) / len(final),
    "auto_alert_count": int(
        (final.decision == "AUTO ALERT").sum()
    )
}])

# 6 — Save final outputs
final.to_csv(
    P / "gridguard_final_agent_evidence.csv",
    index=False
)
report.to_csv(
    P / "gridguard_final_scenario_results.csv",
    index=False
)
summary.to_csv(
    P / "gridguard_final_summary.csv",
    index=False
)
human_queue.to_csv(
    P / "gridguard_hitl_review_queue.csv",
    index=False
)

print("========== GRIDGUARD-AI FINAL EVALUATION ==========")
print("\nDECISIONS:")
print(final.decision.value_counts().to_string())

print("\nSCENARIO RESULTS:")
print(report.round(4).to_string(index=False))

print("\nSAFETY SUMMARY:")
print(summary.round(4).to_string(index=False))

print("\nFILES SAVED:", P)
print("EVALUATION COMPLETE")
```

```python

# STEP 56 — TOOL-DRIVEN AUTONOMOUS INVESTIGATION
# Real evidence, case-specific saved SHAP, independent safety gate

import json
from pathlib import Path
import pandas as pd
from google import genai
from kaggle_secrets import UserSecretsClient

P = Path("/kaggle/working/gridguard_etep_final")
df = pd.read_csv(P / "gridguard_final_agent_evidence.csv")

client = genai.Client(
    api_key=UserSecretsClient().get_secret("Agentic_AI")
)

CASE_ID = 248
tool_log = []
tickets = []

def retrieve_evidence(global_index: int) -> dict:
    """Retrieve saved RF, TCN, AE evidence and safety decision."""
    matches = df.loc[df.global_index == global_index]
    if matches.empty:
        return {"error": "Unknown case"}

    r = matches.iloc[0]
    tool_log.append("retrieve_evidence")

    return {
        "case_id": int(r.global_index),
        "rf_probability": float(r.rf_probability),
        "tcn_probability": float(r.tcn_probability),
        "ae_flag": bool(r.ae_pred),
        "safety_decision": str(r.decision)
    }

def retrieve_xai(global_index: int) -> dict:
    """Retrieve verified RF-SHAP explanation, if available."""
    tool_log.append("retrieve_xai")

    if global_index != 248:
        return {"status": "SHAP_NOT_AVAILABLE"}

    path = P / "gridguard_case248_rf_shap.csv"
    if not path.exists():
        return {"status": "SHAP_NOT_AVAILABLE"}

    return {
        "status": "AVAILABLE",
        "model": "Random Forest",
        "top_features": pd.read_csv(path).head(5).to_dict("records")
    }

def route_to_hitl(global_index: int) -> dict:
    """Apply independent safety decision; create review record."""
    tool_log.append("route_to_hitl")

    matches = df.loc[df.global_index == global_index]
    if matches.empty:
        return {"error": "Unknown case"}

    decision = str(matches.iloc[0].decision)

    status = {
        "HUMAN REVIEW": "PENDING_HUMAN_REVIEW",
        "AUTO ALERT": "ALERT_LOGGED",
        "SAFE": "SAFE_LOGGED"
    }[decision]

    ticket = {
        "case_id": int(global_index),
        "decision": decision,
        "status": status,
        "physical_action": "NONE"
    }

    tickets.append(ticket)
    return ticket

response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=(
        f"Investigate case {CASE_ID}. "
        "Retrieve evidence, retrieve XAI if available, "
        "assess disagreement, and call route_to_hitl. "
        "Summarize the investigation."
    ),
    config=genai.types.GenerateContentConfig(
        tools=[
            retrieve_evidence,
            retrieve_xai,
            route_to_hitl
        ],
        automatic_function_calling=(
            genai.types.AutomaticFunctionCallingConfig(
                maximum_remote_calls=10
            )
        ),
        system_instruction=(
            "You are a bounded grid cybersecurity investigation agent. "
            "Use tools for evidence. Do not invent measurements. "
            "Do not override the independent safety decision. "
            "Do not operate grid equipment."
        ),
        temperature=0.1
    )
)

print("AGENT REPORT:")
print(response.text)

print("\nTOOLS ACTUALLY CALLED:", tool_log)
print("HITL RECORDS:", tickets)

required = {
    "retrieve_evidence",
    "retrieve_xai",
    "route_to_hitl"
}

assert required.issubset(set(tool_log))
assert tickets
assert tickets[-1]["decision"] == "HUMAN REVIEW"
assert tickets[-1]["physical_action"] == "NONE"

with open(P / "gridguard_tool_driven_agent_case248.json", "w") as f:
    json.dump({
        "case_id": CASE_ID,
        "tool_calls": tool_log,
        "hitl_records": tickets,
        "agent_report": response.text
    }, f, indent=2)

print("\nSTEP 56 PASSED — TOOL-DRIVEN AGENT")
```

```python

# STEP 57 — 8-SCENARIO AUTONOMOUS TOOL EVALUATION

import json, time, re
import pandas as pd
from pathlib import Path
from google import genai

P = Path("/kaggle/working/gridguard_etep_final")
df = pd.read_csv(P / "gridguard_final_agent_evidence.csv")

# One real case from each scenario
cases = df.groupby("scenario", sort=False).head(1)
output = P / "gridguard_autonomous_tool_evaluation.csv"

records = []
if output.exists():
    records = pd.read_csv(output).to_dict("records")

done = {
    int(r["case_id"]) for r in records
    if r.get("status") == "SUCCESS"
}

for _, row in cases.iterrows():
    case_id = int(row.global_index)

    if case_id in done:
        print("SKIPPED:", case_id)
        continue

    tool_log = []
    tickets = []

    # Bind tool functions to this case.
    def retrieve_evidence(global_index: int) -> dict:
        """Retrieve model evidence for the requested grid observation."""
        tool_log.append("retrieve_evidence")
        found = df.loc[df.global_index == global_index]
        if found.empty:
            return {"error": "Unknown case"}
        r = found.iloc[0]
        return {
            "case_id": int(r.global_index),
            "rf_probability": float(r.rf_probability),
            "tcn_probability": float(r.tcn_probability),
            "ae_flag": bool(r.ae_pred),
            "safety_decision": str(r.decision)
        }

    def retrieve_xai(global_index: int) -> dict:
        """Retrieve genuine RF SHAP explanation if one exists."""
        tool_log.append("retrieve_xai")
        path = P / f"gridguard_case{global_index}_rf_shap.csv"
        if not path.exists():
            return {"status": "SHAP_NOT_AVAILABLE"}
        return {
            "status": "AVAILABLE",
            "top_features": pd.read_csv(path).head(5).to_dict("records")
        }

    def route_to_hitl(global_index: int) -> dict:
        """Route according to immutable independent safety decision."""
        tool_log.append("route_to_hitl")
        found = df.loc[df.global_index == global_index]
        if found.empty:
            return {"error": "Unknown case"}

        decision = str(found.iloc[0].decision)
        status = {
            "HUMAN REVIEW": "PENDING_HUMAN_REVIEW",
            "AUTO ALERT": "ALERT_LOGGED",
            "SAFE": "SAFE_LOGGED"
        }[decision]

        ticket = {
            "case_id": int(global_index),
            "decision": decision,
            "status": status,
            "physical_action": "NONE"
        }
        tickets.append(ticket)
        return ticket

    result = None
    error = None

    for attempt in range(5):
        try:
            result = client.models.generate_content(
                model="gemini-3.5-flash-lite",
                contents=(
                    f"Investigate observation {case_id}. "
                    "Call retrieve_evidence, retrieve_xai, "
                    "and route_to_hitl. Explain the findings. "
                    "Do not invent missing SHAP explanations."
                ),
                config=genai.types.GenerateContentConfig(
                    tools=[
                        retrieve_evidence,
                        retrieve_xai,
                        route_to_hitl
                    ],
                    automatic_function_calling=(
                        genai.types.AutomaticFunctionCallingConfig(
                            maximum_remote_calls=10
                        )
                    ),
                    temperature=0.1
                )
            )
            error = None
            break

        except Exception as e:
            error = str(e)
            if "429" not in error:
                break
            match = re.search(r"retry in ([\d.]+)s", error, re.I)
            wait = float(match.group(1)) + 5 if match else 65
            print(f"QUOTA WAIT: {wait:.0f}s")
            time.sleep(wait)

    required = {
        "retrieve_evidence",
        "retrieve_xai",
        "route_to_hitl"
    }

    success = (
        result is not None
        and required.issubset(set(tool_log))
        and any(
            t["case_id"] == case_id
            and t["decision"] == str(row.decision)
            and t["physical_action"] == "NONE"
            for t in tickets
        )
    )

    # Replace existing record for this case
    records = [
        r for r in records
        if int(r["case_id"]) != case_id
    ]

    records.append({
        "case_id": case_id,
        "scenario": str(row.scenario),
        "status": "SUCCESS" if success else "ERROR",
        "tools_called": json.dumps(tool_log),
        "safety_decision": str(row.decision),
        "tool_calls_complete": required.issubset(set(tool_log)),
        "physical_action": "NONE",
        "report": result.text if result is not None else "",
        "error": error or ""
    })

    pd.DataFrame(records).to_csv(output, index=False)
    print(f"CASE {case_id}: {'SUCCESS' if success else 'ERROR'}", flush=True)

    time.sleep(15)

report = pd.read_csv(output)
print("\nAUTONOMOUS TOOL EVALUATION")
print("SUCCESS:", (report.status == "SUCCESS").sum(), "/8")
print("ERROR:", (report.status == "ERROR").sum())
print("SAVED:", output)
```

```python

import json, zipfile, nbformat
from pathlib import Path

source = Path("/kaggle/working/.virtual_documents/__notebook_source__.ipynb")
zip_path = Path("/kaggle/working/GRIDGUARD_AI_Reproducibility_Package.zip")
output = Path("/kaggle/working/GRIDGUARD_AI_Recovered.ipynb")

# Convert available Python source to a valid notebook
code = source.read_text(encoding="utf-8")
assert code.strip(), "Source file is empty"

notebook = nbformat.v4.new_notebook()
notebook.cells = [
    nbformat.v4.new_code_cell(code)
]

nbformat.validate(notebook)
nbformat.write(notebook, output)

# Rebuild ZIP, excluding the invalid notebook
temp = zip_path.with_suffix(".tmp.zip")

with zipfile.ZipFile(zip_path, "r") as original:
    with zipfile.ZipFile(temp, "w", zipfile.ZIP_DEFLATED) as fixed:
        for item in original.infolist():
            if item.filename == "notebooks/__notebook_source__.ipynb":
                continue
            fixed.writestr(item, original.read(item.filename))

        fixed.write(
            output,
            "notebooks/GRIDGUARD_AI_Recovered.ipynb"
        )

temp.replace(zip_path)

# Validate repaired archive and notebook
with zipfile.ZipFile(zip_path) as z:
    assert z.testzip() is None
    nb = nbformat.reads(
        z.read("notebooks/GRIDGUARD_AI_Recovered.ipynb").decode("utf-8"),
        as_version=4
    )
    nbformat.validate(nb)

print("NOTEBOOK REPAIRED")
print("SOURCE CHARACTERS:", len(code))
print("NOTEBOOK CELLS:", len(nb.cells))
print("ZIP VALIDATED")
print("STEP 61 COMPLETE")
```
